```python
smiles = []
for index, row in result.iterrows():
    try:
        mol = Chem.MolFromInchi(row['inchi'])
    except Exception as e:
        print(e)
        mol = None

    if mol:
        try:
            smile = Chem.MolToSmiles(mol, canonical=True)
        except Exception as e:
            print(e)
            smile = None
    else:
        smile = None
    smiles.append(np.nan if smile is None else smile)
result['smiles'] = smiles
```

```python
result = result.dropna()
```
